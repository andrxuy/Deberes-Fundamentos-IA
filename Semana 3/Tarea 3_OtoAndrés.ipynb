```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Cargar dataset
df = pd.read_csv("Fortnite Statistics.csv")

# Definir Variable Dependiente (Y)
Y = df['Eliminations']
```

```python
#3. El coeficiente de correlación de una variable independiente y otra dependiente (aquellas que usted haya seleccionado para trabajar)
# Seleccionar Variable Independiente 1 
X_simple = df[['Hits']]

# Calcular Coeficiente de Pearson (r)
r_pearson = np.corrcoef(X_simple.values.flatten(), Y.values)[0, 1]

print(f'3. El coeficiente de correlación de una variable independiente y otra dependiente (aquellas que usted haya seleccionado para trabajar)')
print(f"Coeficiente de Correlación de Pearson (r) [Hits vs Eliminations]: {r_pearson:.4f}")
```

```python
#4. Los valores de las métricas estudiadas en clase MSE, MAE
# Dividir datos para el modelo simple
X_train_s, X_test_s, Y_train_s, Y_test_s = train_test_split(X_simple, Y, test_size=0.2, random_state=42)

# Entrenar modelo simple
modelo_simple = LinearRegression()
modelo_simple.fit(X_train_s, Y_train_s)

# Predecir
Y_pred_s = modelo_simple.predict(X_test_s)

# Calcular MSE y MAE
mse = mean_squared_error(Y_test_s, Y_pred_s)
mae = mean_absolute_error(Y_test_s, Y_pred_s)

print(f"4. Los valores de las métricas estudiadas en clase MSE, MAE")
print(f"MSE (Error Cuadrático Medio): {mse:.4f}")
print(f"MAE (Error Absoluto Medio): {mae:.4f}")
```

```python
#4. Los valores de las métricas estudiadas en clase MSE, MAE
# Dividir datos para el modelo simple
X_train_s, X_test_s, Y_train_s, Y_test_s = train_test_split(X_simple, Y, test_size=0.2, random_state=42)

# Entrenar modelo simple
modelo_simple = LinearRegression()
modelo_simple.fit(X_train_s, Y_train_s)

# Predecir
Y_pred_s = modelo_simple.predict(X_test_s)

# Calcular MSE y MAE
mse = mean_squared_error(Y_test_s, Y_pred_s)
mae = mean_absolute_error(Y_test_s, Y_pred_s)

print(f"4. MSE, MAE")
print(f"MSE (Error Cuadrático Medio): {mse:.4f}")
print(f"MAE (Error Absoluto Medio): {mae:.4f}")
```

```python
#5. Los valores de las tres métricas de evaluación consultadas (RMSE, R² y R² ajustado)

# Seleccionar Variables Independientes (X1 y X2)
X_multi = df[['Hits', 'Damage to Players']]

# Dividir datos para el modelo múltiple
X_train_m, X_test_m, Y_train_m, Y_test_m = train_test_split(X_multi, Y, test_size=0.2, random_state=42)

# Entrenar modelo múltiple
modelo_multi = LinearRegression()
modelo_multi.fit(X_train_m, Y_train_m)

# Predecir
Y_pred_m = modelo_multi.predict(X_test_m)

# --- MÉTRICAS PUNTO 5 ---
mse_m = mean_squared_error(Y_test_m, Y_pred_m)
rmse = np.sqrt(mse_m)
r2 = r2_score(Y_test_m, Y_pred_m)

# Cálculo del R2 Ajustado
n = len(Y_test_m)      # Número de muestras de prueba
k = X_multi.shape[1]   # Número de variables independientes (2)
r2_ajustado = 1 - (1 - r2) * (n - 1) / (n - k - 1)


print(f"5. RMSE, R² y R² ajustado")
print(f"RMSE (Raíz del Error Cuadrático Medio): {rmse:.4f}")
print(f"R2 (Coeficiente de Determinación): {r2:.4f}")
print(f"R2 Ajustado: {r2_ajustado:.4f}")

```

```python
#6. El valor del coeficiente de correlación de dos variables independientes y una dependiente.

R_multiple = np.sqrt(r2)
print(f"6.El valor del coeficiente de correlación ")
print(f"Coeficiente de Correlación Múltiple (R) [Hits + Damage vs Eliminations]: {R_multiple:.4f}")
```
